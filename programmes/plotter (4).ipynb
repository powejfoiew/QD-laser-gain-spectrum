```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
```

```python
df = pd.read_csv(r'C:\Users\josep\Documents\MRes mini-project 2\Figures\Population-Time\GandR_calib.csv',names =['x','y'])

x = df['x']
y = df['y']

plt.plot(x,y)
res = stats.linregress(x, y)
print('slope:', res[0])
print('intercept:', res[1])
```

```python
df2 = pd.read_csv(r'C:\Users\josep\Documents\MRes mini-project 2\Figures\Population-Time\Gand_occucalib.csv',names=['x','y'])
power = df2['x']
occupation = df2['y']
current_from_power = (power-res[1])/res[0]
#print(current_from_power)
plt.scatter(current_from_power,occupation)
plt.show()
df2['z']=current_from_power
df2.to_csv('Rossetti_GS_occupations.csv')
```
